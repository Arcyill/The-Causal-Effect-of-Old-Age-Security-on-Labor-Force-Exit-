```r
library(tidyverse)
library(ggplot2)
library(haven)
library(stargazer)
```

```r
df <- read_dta("data/cen_ind_2021_pumf_v2.dta")
head(df)
```

```r
names(df)
```

```r
selected_df <- df|>
    select(lfact, agegrp, immstat, ageimm, yrim, hdgree, Gender, marsth, vismin) 
head(selected_df)
```

```r
summary(selected_df)
```

```r
filtered_df <- selected_df|>
    filter(lfact < 14) |>
    filter(between(agegrp, 16, 17)) |>
    filter(immstat < 3) |> #Remove non-permanent residents and not available
    filter(ageimm != 88) |> #keep 99 here since that's canadian born
    filter(yrim != 8888) |> #keep 9999 here since that's canadian born
    filter(vismin != 88) |>

    mutate(eligible = if_else(immstat == 1 | (immstat == 2 & yrim <= 2011), 1, 0)) |>
    mutate(NILF = if_else(lfact >= 11, 1, 0)) |>
    mutate(agegrp = if_else(agegrp == 16, 0, 1)) |> #Change age group to dummy, 0 if 16, 1 if 17
    mutate(women = if_else(Gender == 1, 1, 0)) |>
    mutate(vismin_dummy = if_else(vismin > 1, 1, 0)) |>

    #categorical variables
    filter(hdgree != 88 & hdgree != 99) |> 
    mutate(educ_level = case_when(
        hdgree == 1 ~ "1_Less_Than_HS",
        hdgree == 2 ~ "2_High_School",
        hdgree >= 3 & hdgree <= 8 ~ "3_College_Trades",
        hdgree >= 9 & hdgree <= 13 ~ "4_University",
        TRUE ~ NA_character_
    )) |> 
    mutate(educ_level = as.factor(educ_level)) |>

    filter(marsth != 8) |> 
    mutate(marital_group = case_when(
        marsth %in% c(2, 3) ~ "1_Partnered",         # Code 2 (Married), Code 3 (Common law)
        marsth %in% c(4, 5, 6) ~ "2_Prev_Married",   # Code 4 (Separated), 5 (Divorced), 6 (Widowed)
        marsth == 1 ~ "3_Never_Married",                    # Code 1 (Never married)
        TRUE ~ NA_character_
    )) |>
    mutate(marital_group = as.factor(marital_group)) |>

    select(NILF, eligible, agegrp, women, vismin_dummy, educ_level, marital_group)

head(filtered_df)
```

```r
summary(filtered_df)
```

before running regressions, it is worth checking if the treatment group 

```r
t.test(vismin_dummy ~ eligible, data = filtered_df)
```

```r
base_reg <- lm(NILF ~ eligible, data = filtered_df)
summary(base_reg)
```

```r
stargazer(base_reg, type = "text")
```

```r
full_reg <- lm(NILF ~ eligible + agegrp ++ women + educ_level + marital_group + vismin_dummy, data = filtered_df)
summary(full_reg)
```

```r
did_cov <- lm(NILF ~ eligible + agegrp + eligible * agegrp, data = filtered_df)

summary(did_cov)
```

```r
stargazer(did_cov, type = "text", title = "DiD Baseline No Controls")
```

```r
did_cov_ctrl <- lm(NILF ~ eligible + agegrp + eligible * agegrp + women + educ_level + marital_group + vismin_dummy, data = filtered_df)

summary(did_cov_ctrl)
```

```r
stargazer(did_cov_ctrl, type = "text", title = "DiD with Controls")

```

```r
install.packages("webshot2")
library(webshot2)

stargazer(
    base_reg, full_reg, did_cov, did_cov_ctrl,
    type = "text", #changed the type here
    title = "Cross-sectional DD-style models",
    dep.var.labels = "Not in labour force (NILF)",
    column.labels = c("Baseline", "Controls", "DiD", "DiD + controls"),
    covariate.labels = c(
        "Eligible (OAS)",
        "Age 65-69",
        "Female",
        "High school (ref: < HS)",
        "College/Trades (ref: < HS)",
        "University (ref: < HS)",
        "Previously married (ref: partnered)",
        "Never Married (ref: partnered)",
        "Visible minority",
        "Eligible x Age 65-69"),
    digits = 3,
    out = "dd_style_table_full.txt" #file created
)

webshot("dd_style_table_full.txt", "dd_style_table_full.png",vwidth = 1200, vheight = 800)
```
